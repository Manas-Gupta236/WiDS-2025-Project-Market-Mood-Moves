```python
import pandas as pd

raw_sentiment = pd.read_csv("raw_sentiment.csv")
raw_sentiment["date"] = pd.to_datetime(raw_sentiment["date"]).dt.date

raw_sentiment["sentiment_score"] = (
    raw_sentiment["positive"] - raw_sentiment["negative"]
)

daily_sentiment = (
    raw_sentiment
    .groupby("date")["sentiment_score"]
    .mean()
    .reset_index()
)

daily_sentiment.to_csv("daily_sentiment.csv", index=False)
daily_sentiment
```
