```python
import torch
import pandas as pd
import numpy as np
from torch.utils.data import DataLoader
import torch.optim as optim
import torch.nn as nn
import matplotlib.pyplot as plt

from dataset import MarketDataset
from lstm_model import SentimentLSTM

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Load data
df = pd.read_csv("../final_week3_dataset.csv")

features = [
    'sentiment_lag1',
    'sentiment_3d',
    'sentiment_7d',
    'volatility_5d',
    'return'
]

X = df[features].values
y = df['return'].values

dataset = MarketDataset(X, y, seq_len=60)

train_size = int(0.8 * len(dataset))
train_set, test_set = torch.utils.data.random_split(dataset, [train_size, len(dataset) - train_size])

train_loader = DataLoader(train_set, batch_size=32, shuffle=False)
test_loader = DataLoader(test_set, batch_size=32, shuffle=False)

model = SentimentLSTM(input_dim=len(features)).to(device)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# Training
losses = []
for epoch in range(30):
    model.train()
    for Xb, yb in train_loader:
        Xb, yb = Xb.to(device), yb.to(device)

        optimizer.zero_grad()
        preds = model(Xb).squeeze()
        loss = criterion(preds, yb)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()

    losses.append(loss.item())
    print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")

# Evaluation
model.eval()
actuals, predictions = [], []

with torch.no_grad():
    for Xb, yb in test_loader:
        Xb = Xb.to(device)
        pred = model(Xb).cpu().numpy()
        predictions.extend(pred.flatten())
        actuals.extend(yb.numpy())

plt.figure()
plt.plot(actuals, label="Actual")
plt.plot(predictions, label="Predicted")
plt.legend()
plt.title("LSTM Prediction vs Actual")
plt.show()

```
